```python
import math
import numpy as np

x = np.linspace(0, 2 * math.pi, 200)
y = np.sin(x) * np.exp(-0.15 * x)
print(f"Computed {len(x)} points, peak = {y.max():.4f}")
y[:5]
```

## Inline Matplotlib Plots

In **Ghostty** and **Kitty**, plots render natively inline using Kitty Graphics Unicode Placeholders (press `<leader>ji` to open full-size externally):

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 3.2), dpi=120)
ax.plot(x, y, color="#89b4fa", linewidth=2.2, label="Damped sine")
ax.fill_between(x, y, alpha=0.18, color="#89b4fa")
ax.set_title("Damped Harmonic Oscillation")
ax.set_xlabel("t (rad)")
ax.set_ylabel("Amplitude")
ax.grid(True, alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()
```
